ΠΡΩΤΗ ΕΠΑΦΗ ΜΕ ΤΑ DATA ΜΑΣ ΠΡΙΝ ΤΟ PREPROCESSING

```python
import pandas as pd
import glob
import os

current_dir = os.getcwd()

path = os.path.join(current_dir, "..", "data", "raw")
all_files = glob.glob(os.path.join(path,"*.csv"))

print("Loading Data")

df = pd.concat((pd.read_csv(f, low_memory=False) for f in all_files), ignore_index=True)
    
df.columns = df.columns.str.strip()




```

ΑΝΑΛΥΣΗ ΤΟΥ ΚΑΘΕ DATASET ΞΕΧΩΡΙΣΤΑ

```python
import os

list_df = []

for f in all_files:
    temp_df = pd.read_csv(f, low_memory=False)
    temp_df.columns = temp_df.columns.str.strip()
    
    # this is a new col with the file name 
    temp_df['source_file'] = os.path.basename(f) 
    
    list_df.append(temp_df)

df = pd.concat(list_df, ignore_index=True)

print(df.groupby('source_file')['Label'].value_counts())
```
